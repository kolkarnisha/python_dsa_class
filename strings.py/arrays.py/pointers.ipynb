```python
''' day 37 date 13-10-26  
longest substring with distinct characters'''
class S:
    def c(self,s):
        freq=[-1]*26
        res=0
        start=0
        for i in range(len(s)):
            start=max(start,freq[ord(s[i])-ord("a")]+1)
            res=max(res,i-start+1)
            freq[ord(s[i])-ord("a")]=i
        return res
s="geeksforgeek"
o=S()
o.c(s)
```
