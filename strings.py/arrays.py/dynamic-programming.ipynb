```python
'''day 48'''
'''date 04-11-26'''
''' Palindrome Substrings Count'''
'''abaaba
low=i
high=i+1
low=high-1
high=i+1 '''
def c(s):
    count=0
    n=len(s)
    for i in range(n):
        low=i
        high=i+1
        while low>=0 and high<n and s[low]==s[high]:     #even number of characters
            count+=1
            low-=1
            high+=1
        low=i-1
        high=i+1
        while low>=0 and high<n and s[low]==s[high]:   #odd number of characters
            count+=1
            low-=1
            high+=1
    return count
s='abaab'
c(s)
```
